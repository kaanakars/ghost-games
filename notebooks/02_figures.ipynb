# 02 · Final charts

**Question:** How much of football's home advantage disappeared when stadiums were empty, and through which channel?

Builds the three README charts and saves them to `figures/`. Re-run after any change to the data or the model.

```python
%load_ext autoreload
%autoreload 2

from ghost_games import figures
from ghost_games.pipeline import build_matches

matches = build_matches("..")
```

```python
fig = figures.plot_over_time(matches)
fig.savefig("../figures/over_time.png", dpi=200, facecolor=figures.SURFACE)
# Takeaway: the home goal edge fell when stadiums emptied in 2020 and came back with the fans.
```

```python
fig = figures.plot_share_lost(matches)
fig.savefig("../figures/share_lost.png", dpi=200, facecolor=figures.SURFACE)
# Takeaway: every measure lost most of its home edge, 60-68% for goals and points,
# and the referee's card edge disappeared completely.
```

```python
fig = figures.plot_by_league(matches)
fig.savefig("../figures/by_league.png", dpi=200, facecolor=figures.SURFACE)
# Takeaway: the card edge fell in all five leagues; the goal edge fell everywhere,
# but Serie A's drop is within its margin of error.
```
